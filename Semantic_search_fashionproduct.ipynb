```python
import pandas as pd
fashion= pd.read_csv("Styles.csv")
```

```python
print(fashion.head())
```

```python
fashion.shape
```

```python
fashion=fashion.drop(columns=['Unnamed: 4'])
```

```python
fashion["link"] = fashion["link"].str.replace(
    "http://",
    "https://",
    regex=False
)
```

```python
fashion.head()
```

```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer('all-MiniLM-L6-v2')
sample = fashion['Product Description'].fillna("").to_list()
embeddings = model.encode(sample)
print(embeddings.shape)
```

```python
!pip install chromadb
```

```python
import chromadb
client = chromadb.PersistentClient(path="./fashion_db")
collection= client.create_collection("Fashion_Styles1")

batch_size= 5000
for i in range(0,len(sample), batch_size):
  collection.add(
    documents= sample[i:i+batch_size],
    embeddings = embeddings[i:i+batch_size].tolist(),
    ids=[str(i) for i in fashion.index[i:i+batch_size]]
)
```

```python
query_embedding = model.encode(["black track pants"]).tolist()
results = collection.query(query_embeddings=query_embedding, n_results=5)
print(results['documents'])
```

```python
query_embedding = model.encode(["summer floral dress"]).tolist()
results = collection.query(query_embeddings=query_embedding, n_results=5)
print(results['documents'])
```

```python
%%writefile app.py

import streamlit as st
import pandas as pd
import chromadb
fashion=pd.read_csv("Styles.csv")
fashion["link"] = fashion["link"].str.replace(
    "http://",
    "https://",
    regex=False
)



client = chromadb.PersistentClient(path="./fashion_db")
collection = client.get_collection("Fashion_Styles1")

st.title("Fashion Search")

query = st.text_input("Search for a product")

if query:
    results = collection.query(
        query_texts=[query],
        n_results=5
    )

    ids = results["ids"][0]

    for id in ids:

      product = fashion.loc[int(id)]

      with st.container():

        col1, col2 = st.columns([1,3])

        with col1:

            if "link" in fashion.columns:
                st.image(product["link"], width=140)

        with col2:

            st.markdown(f"### {product['Product Description']}")
            st.write(f"**Category:** {product['Category']}")


        st.divider()
```
